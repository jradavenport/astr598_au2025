```python
# you'll need at least these packages today
import numpy as np
import matplotlib.pyplot as plt

import lightkurve as lk
from astropy.timeseries import LombScargle, BoxLeastSquares
import batman
```

```python
# HERE ARE OUR EB's:

# KIC 11097678
# KIC 10346522
# KIC 10651945
# KIC 10215422

# you can look at versions of these systems from the Villanova EB webpage
#  e.g. https://keplerebs.villanova.edu/overview/?k=10346522
```

```python
# lightkurve is the go-to handy package for finding and using Kepler and TESS data. 
# It can do pixel-level (near raw) data, or processed light curves (our goal today)
lk.search_lightcurve('KIC 10651945', author='Kepler')
```

```python
LC = lk.search_lightcurve('KIC 10651945', author='Kepler').download_all()
```

```python
LC.plot() # a quick helper function to plot a lightcurve collection
```

```python
LC.stitch().flatten(window_length=201).plot()
plt.show()
```
