```python
# MANY different source models to draw from... check it out: https://sncosmo.readthedocs.io/en/stable/source-list.html

# 'salt2' is a standard one, as is 'hsiao'
model = sncosmo.Model('salt2')

# fix some params for redshift (has to be >0) and event peak time
model.set(z=0.01, t0=100.0)

# set the peak brightess... for SOME models this helps, sometimes is behaves unintuitively
# might want to skip this depending on your particular model... it's optional, YMMV
model.set_source_peakabsmag(-19.36, 'ztfr', 'ab')

# generate light curves in different bands
tt= np.arange(90,200,.1)
plt.figure()

# For brute force alignment, you might want to find the model peak brightness and just shift it to match our data
plt.plot(tt, model.bandmag('ztfg', 'ab', tt))
plt.plot(tt, model.bandmag('ztfr', 'ab', tt))
# NOTE: some model might return NaN's or INF's outside of time range it can simulate (way before or way after peak)
plt.gca().invert_yaxis()
```
